# Notebook Overview: 00_setup

## Purpose
This notebook sets up the Unity Catalog infrastructure for the Olist Databricks Project by creating a catalog and establishing a medallion architecture for data organization.

## Structure & Operations

### Catalog Setup
* **Creates catalog**: `olist_dataset` (Cell 3)
* **Activates catalog**: Sets `olist_dataset` as the current catalog (Cell 4)

### Schema Architecture
Creates a multi-layer data architecture within the `olist_dataset` catalog (Cell 5):

| Schema | Purpose |
| --- | --- |
| `raw_dataset` | Likely for raw, unprocessed data |
| `bronze` | Raw data ingestion layer |
| `silver` | Cleaned and validated data |
| `gold` | Business-level aggregates and analytics |

### Verification
* **Cell 6**: Lists all schemas in the catalog to confirm successful creation

```python
%sql 
CREATE CATALOG IF NOT EXISTS olist_dataset
```

```python
%sql
USE CATALOG olist_dataset
```

```python
%sql 
CREATE SCHEMA IF NOT EXISTS raw_dataset;
CREATE SCHEMA IF NOT EXISTS bronze;
CREATE SCHEMA IF NOT EXISTS silver;
CREATE SCHEMA IF NOT EXISTS gold;
```
